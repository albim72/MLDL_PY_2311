```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import kmeans_plusplus
```

```python
n_samples = 45_000
n_components = 8
```

```python
X,y_true = make_blobs(n_samples=n_samples, centers=n_components, random_state=1,cluster_std=0.4)
X = X[:,::-1]
```

```python
centers_init,indices = kmeans_plusplus(X,n_clusters=8,random_state=42)
```

```python
plt.figure()
colors = ["#4EACC5","#DD94C6","m","c","#AA567E","#1111EE","y","g","#453AA2"]
for k,col in enumerate(colors):
  cluster_data = y_true == k
  plt.scatter(X[cluster_data,0],X[cluster_data,1],c=col,marker=".",s=10)
plt.scatter(centers_init[:,0],centers_init[:,1],c="r",marker="+",s=100)
plt.title("algorytm k-means")
plt.xticks([])
plt.yticks([])
plt.show()
```
